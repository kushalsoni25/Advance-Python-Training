```python
class A:
  def showA(self):
    print("ShowA function")

class B:
  def showB(self):
    print("ShowB function")

class C(A,B):
  def showC(self):
    print("ShowC function")


obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
#method over writting

class Parent:
  def bike(self):
    print("Pulsar 150")

class Child(Parent):
  def bike(self):
    print("Duke 390 R")

obj=Child()
obj.bike()
```
